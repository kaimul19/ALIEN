```python

torch.set_float32_matmul_precision('high')

def final_loss(recon_loss, mu, logvar,
               beta: float = 1.0,
               lambda_offdiag: float = 10.0,
               lambda_diag: float = 5.0):
    """
    recon_loss: mean‐reduced reconstruction loss (e.g. BCE or MSE)
    mu, logvar: tensors of shape (B, latent_dim)
    beta: weight on the KL divergence
    lambda_offdiag, lambda_diag: DIP‐VAE covariance penalties
    """
    # KL term (mean over batch)
    kld_elem = 1 + logvar - mu.pow(2) - logvar.exp()       # (B, D)
    KLD = -0.5 * torch.mean(torch.sum(kld_elem, dim=1))    # scalar

    # DIP‐VAE penalty on the covariance of mu
    B, D = mu.shape
    mu_centered = mu - mu.mean(dim=0, keepdim=True)        # (B, D)
    cov = (mu_centered.t() @ mu_centered) / B              # (D, D)

    off_diag = cov - torch.diag(torch.diagonal(cov))
    loss_off = torch.sum(off_diag**2)

    diag_dev = torch.diagonal(cov) - 1.0
    loss_diag = torch.sum(diag_dev**2)

    dip_penalty = lambda_offdiag * loss_off + lambda_diag * loss_diag

    return recon_loss + beta * KLD + dip_penalty


def train(model, dataloader, device, optimizer, criterion):
    model.train()
    running_loss = 0.0
    for batch in tqdm(dataloader, desc='Train'):
        # unpack if DataLoader returns (input, label)
        data = batch[0] if isinstance(batch, (tuple, list)) else batch
        data = data.to(device)

        optimizer.zero_grad()
        reconstruction, mu, logvar = model(data)
        recon_loss = criterion(reconstruction, data)
        loss = final_loss(recon_loss, mu, logvar)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    return running_loss / len(dataloader)


def validate(model, dataloader, device, criterion):
    model.eval()
    running_loss = 0.0
    recon_images = None
    total_batches = len(dataloader)

    with torch.no_grad():
        for i, batch in enumerate(tqdm(dataloader, desc='Valid')):
            data = batch[0] if isinstance(batch, (tuple, list)) else batch
            data = data.to(device)

            reconstruction, mu, logvar = model(data)
            recon_loss = criterion(reconstruction, data)
            loss = final_loss(recon_loss, mu, logvar)
            running_loss += loss.item()

            # Save last batch of reconstructions
            if i == total_batches - 1:
                recon_images = reconstruction

    return running_loss / total_batches, recon_images
```

```python
import numpy as np

import torch
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split

device = torch.device('cuda:2' if torch.cuda.is_available() else 'cpu')

# set the learning parameters
lr = 0.01
epochs = 20
batch_size = 256

grid_images = []  # for visualizations

# Load metadata and data
data_path = "generated_data/background_seperated_processed_data.npy"

shape_path = data_path.replace('.npy', '_shape.npy')
metadata_path = data_path.replace('.npy', '_injection_metadata.npy')


shape = tuple(np.load(shape_path).astype(int))

data = np.memmap(data_path, dtype='float32', mode='r', shape=shape)
metadata = np.load(metadata_path, allow_pickle=True)
# N = 77692

# Check for bounding values
# for i, x in tqdm(enumerate(data[:N]), total=N):
#     max_val = x.max().item()
#     min_val = x.min().item()
#     if max_val > 1.0 or min_val < 0.0:
#         print(f"⚠️ Batch {i}: min={min_val}, max={max_val}")
#         break
# else:
#     print("✅ All values in the dataset are within [0, 1]")

used_data = data
used_metadata = metadata

def prepare_labels(labels):
    expanded_labels = []
    True_count_1 = True_count_2 = False_count_1 = False_count_2 = 0
    num_background = num_linear = 0

    for idx, ((injection_type, cadence_label), frame_flags) in enumerate(labels):
        if injection_type == "Background":
            num_background += 1
        elif injection_type == "Linear":
            num_linear += 1

        if cadence_label:
            True_count_1 += 1
        else:
            False_count_1 += 1

        for frame_idx, frame_flag in enumerate(frame_flags):
            expanded_labels.append((injection_type, cadence_label, frame_flag, (idx, frame_idx)))
            if frame_flag:
                True_count_2 += 1
            else:
                False_count_2 += 1

    print(f"Number of background frames: {num_background}")
    print(f"Number of linear frames: {num_linear}")
    print(f"Number of True in first is {True_count_1}, False is {False_count_1}")
    print(f"Number of True in second is {True_count_2}, False is {False_count_2}")

    return np.array(expanded_labels, dtype=object)

X_labels = prepare_labels(used_metadata)

X_labels, _ = train_test_split(
    X_labels,
    train_size=0.25,
    random_state=42,
    stratify=X_labels[:, 2]
)


print("Data Prepared")


dragon_indicies = np.where(X_labels[:, 2] == "Welsh_dragon")        # 1-D array of length N
non_dragon_indicies = np.where(X_labels[:, 2] != "Welsh_dragon")  # 1-D array of length N


X_labels_no_dragons = X_labels[non_dragon_indicies]
X_labels_with_dragons = X_labels[dragon_indicies]
# Split data
train_ratio = 0.8
val_ratio = 0.1
test_ratio = 0.1



#split the training from the non dragons as dont want to train on dragons
print("First split")
X_train_idx, X_temp_idx, y_train, y_temp = train_test_split(
    np.arange(len(X_labels_no_dragons)), X_labels_no_dragons, test_size=1-train_ratio, random_state=42, stratify=X_labels_no_dragons[:, 2]
)

#Split the leftover from training data
print("Second split")
X_val_idx, X_test_idx_no_dragons, y_val, y_test_no_dragons = train_test_split(
    X_temp_idx, y_temp, test_size=0.5, random_state=42, stratify=y_temp[:, 2]
)

#combine the val and temp randomly
# y_val = np.concatenate((y_val_no_dragons), axis=0) 

y_test = np.concatenate((y_test_no_dragons, X_labels_with_dragons), axis=0)


# Custom dataset
class MemmapFrameDataset(Dataset):
    def __init__(self, memmap_data, labels):
        self.memmap_data = memmap_data
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        inj_type, cadence_label, frame_flag, (cad_idx, frame_idx) = self.labels[idx]
        frame = self.memmap_data[cad_idx, frame_idx]  # (16, 4096)
        frame = np.expand_dims(frame, 0)  # Add channel dim
        frame_tensor = torch.tensor(frame, dtype=torch.float32)
        return frame_tensor, frame_flag

# Create datasets
train_dataset = MemmapFrameDataset(used_data, y_train)
val_dataset = MemmapFrameDataset(used_data, y_val)
test_dataset = MemmapFrameDataset(used_data, y_test)

# Create loaders
trainloader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=25, pin_memory=True)
valloader   = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, num_workers=10, pin_memory=True)
testloader  = DataLoader(test_dataset, batch_size=batch_size, shuffle=False, num_workers=4, pin_memory=True)




```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class FrameConvVAE(nn.Module):
    def __init__(self, image_channels: int = 1, latent_dim: int = 16, beta: float = 1.5):
        super().__init__()
        self.latent_dim = latent_dim
        self.beta       = beta

        dens_lay = 512

        # ─── Encoder conv blocks (8× 3×3) ───
        # downsample by stride=2 every other layer
        self.enc1 = nn.Conv2d(image_channels,  16, kernel_size=3, stride=2, padding=1)  # → [B,16, 8,2048]
        self.bn1  = nn.BatchNorm2d(16)
        self.enc2 = nn.Conv2d(16,             16, kernel_size=3, stride=1, padding=1)  # → [B,16, 8,2048]
        self.bn2  = nn.BatchNorm2d(16)

        self.enc3 = nn.Conv2d(16,             32, kernel_size=3, stride=2, padding=1)  # → [B,32, 4,1024]
        self.bn3  = nn.BatchNorm2d(32)
        self.enc4 = nn.Conv2d(32,             32, kernel_size=3, stride=1, padding=1)  # → [B,32, 4,1024]
        self.bn4  = nn.BatchNorm2d(32)

        self.enc5 = nn.Conv2d(32,             64, kernel_size=3, stride=2, padding=1)  # → [B,64, 2, 512]
        self.bn5  = nn.BatchNorm2d(64)
        self.enc6 = nn.Conv2d(64,             64, kernel_size=3, stride=1, padding=1)  # → [B,64, 2, 512]
        self.bn6  = nn.BatchNorm2d(64)

        self.enc7 = nn.Conv2d(64,            128, kernel_size=3, stride=2, padding=1)  # → [B,128,1, 256]
        self.bn7  = nn.BatchNorm2d(128)
        self.enc8 = nn.Conv2d(128,           128, kernel_size=3, stride=1, padding=1)  # → [B,128,1, 256]
        self.bn8  = nn.BatchNorm2d(128)

        # flattened size = 128 × 1 × 256
        self.flatten_dim = 128 * 1 * 256

        # ─── Bottleneck MLP ───
        self.fc_enc1   = nn.Linear(self.flatten_dim, dens_lay)
        self.bn_enc1   = nn.BatchNorm1d(dens_lay)
        self.fc_mu     = nn.Linear(dens_lay, latent_dim)
        self.fc_logvar = nn.Linear(dens_lay, latent_dim)

        # ─── Decoder MLP ───
        self.fc_dec1   = nn.Linear(latent_dim, dens_lay)
        self.bn_dec1   = nn.BatchNorm1d(dens_lay)
        self.fc_dec2   = nn.Linear(dens_lay, self.flatten_dim)
        self.bn_dec2   = nn.BatchNorm1d(self.flatten_dim)

        # ─── Decoder deconv blocks (8× 3×3) ───
        # upsample by stride=2 every other layer, mirror of encoder
        self.dec1 = nn.ConvTranspose2d(128,  64, kernel_size=3, stride=2, padding=1, output_padding=1)  # → [B,64, 2,512]
        self.bn_d1 = nn.BatchNorm2d(64)
        self.dec2 = nn.ConvTranspose2d(64,   64, kernel_size=3, stride=1, padding=1)                # → [B,64, 2,512]
        self.bn_d2 = nn.BatchNorm2d(64)

        self.dec3 = nn.ConvTranspose2d(64,   32, kernel_size=3, stride=2, padding=1, output_padding=1)  # → [B,32, 4,1024]
        self.bn_d3 = nn.BatchNorm2d(32)
        self.dec4 = nn.ConvTranspose2d(32,   32, kernel_size=3, stride=1, padding=1)                # → [B,32, 4,1024]
        self.bn_d4 = nn.BatchNorm2d(32)

        self.dec5 = nn.ConvTranspose2d(32,   16, kernel_size=3, stride=2, padding=1, output_padding=1)  # → [B,16, 8,2048]
        self.bn_d5 = nn.BatchNorm2d(16)
        self.dec6 = nn.ConvTranspose2d(16,   16, kernel_size=3, stride=1, padding=1)                # → [B,16, 8,2048]
        self.bn_d6 = nn.BatchNorm2d(16)

        # final upsample to [B,1,16,4096]
        self.dec7 = nn.ConvTranspose2d(16,    1, kernel_size=3, stride=2, padding=1, output_padding=1)  # → [B,1,16,4096]

    @staticmethod
    def reparameterize(mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def forward(self, x):
        # ── Encoder ──
        x = F.relu(self.bn1(self.enc1(x)))
        x = F.relu(self.bn2(self.enc2(x)))
        x = F.relu(self.bn3(self.enc3(x)))
        x = F.relu(self.bn4(self.enc4(x)))
        x = F.relu(self.bn5(self.enc5(x)))
        x = F.relu(self.bn6(self.enc6(x)))
        x = F.relu(self.bn7(self.enc7(x)))
        x = F.relu(self.bn8(self.enc8(x)))
        x = x.view(x.size(0), -1)

        # bottleneck
        h     = F.relu(self.bn_enc1(self.fc_enc1(x)))
        mu    = self.fc_mu(h)
        logvar= self.fc_logvar(h)
        z     = self.reparameterize(mu, logvar)

        # ── Decoder ── MLP → feature map
        h     = F.relu(self.bn_dec1(self.fc_dec1(z)))
        h     = F.relu(self.bn_dec2(self.fc_dec2(h)))
        h     = h.view(-1, 128, 1, 256)

        # deconv path
        h     = F.relu(self.bn_d1(self.dec1(h)))
        h     = F.relu(self.bn_d2(self.dec2(h)))
        h     = F.relu(self.bn_d3(self.dec3(h)))
        h     = F.relu(self.bn_d4(self.dec4(h)))
        h     = F.relu(self.bn_d5(self.dec5(h)))
        h     = F.relu(self.bn_d6(self.dec6(h)))
        logits= self.dec7(h)

        return torch.sigmoid(logits), mu, logvar
```

```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt

import heapq

model.eval()
K = 50

# Full lists for histogram (errors + label strings)
errors_list   = []
labels_list   = []

# Min-heap for storing only the top-K worst (error, orig, recon, label)
heap = []

with torch.no_grad():
    for batch_idx, (data, _) in enumerate(tqdm(testloader, total=len(testloader))):
        B = data.size(0)
        data = data.to(device, non_blocking=True)
        recon, _, _ = model(data)

        # per-sample BCE error
        bce = F.binary_cross_entropy(recon, data, reduction='none')
        per_sample_error = bce.view(B, -1).mean(dim=1).cpu()

        data_cpu  = data.cpu()
        recon_cpu = recon.cpu()

        # compute global index offset for this batch
        start_idx = batch_idx * testloader.batch_size

        for i in range(B):
            err   = per_sample_error[i].item()
            label = y_test[start_idx + i]   # your (inj, cad_lbl, frame_flag, (cad,frm))

            # 1) record for histogram
            errors_list.append(err)
            labels_list.append(label[2])    # or however you want to group

            # 2) maintain heap of full tensors
            orig = data_cpu[i]
            rec  = recon_cpu[i]
            entry = (err, orig, rec, label)

            if len(heap) < K:
                heapq.heappush(heap, entry)
            else:
                if err > heap[0][0]:
                    heapq.heapreplace(heap, entry)

# Extract the worst-K (sorted descending)
worst_items = sorted(heap, key=lambda x: x[0], reverse=True)

# === Plot the worst 50 as before ===
for group in range(10):
    plt.figure(figsize=(12, 6))
    for j in range(5):
        err, orig, rec, label = worst_items[group*5 + j]
        inj, cad_lbl, frame_flag, (cad, frm) = label
        title = f"{inj}, \nCadence {cad}, \nFrame {frm}\nErr={err:.4f}"

        plt.subplot(2, 5, j+1)
        plt.imshow(orig[0], cmap='viridis', aspect='auto')
        plt.title("O\n" + title)
        plt.axis("off")

        plt.subplot(2, 5, j+6)
        plt.imshow(rec[0], cmap='viridis', aspect='auto')
        plt.title("R")
        plt.axis("off")

    plt.tight_layout()
    plt.show()

```

```python
back = 0.1
lin = 1-back

true_split = 0.5
false_split = 1-true_split

number = 100e3

def calc_nums(back_split, lin_split, true_split, false_split, num):
    back_num = back_split * num
    lin_num = lin_split * num
    true_num = lin_num * true_split
    false_num = lin_num * false_split

    num_false_flag = 6*back_num + 3* true_num
    num_true_flag = 3*true_num + 6* false_num

    if round(num_false_flag)+round(num_true_flag)!= round(num*6):
        print(round(num_false_flag)+round(num_true_flag))
        print(round(num*6))
        raise ValueError("The sum of false and true flags does not equal the total number of samples.")

    return round(back_num), round(lin_num), round(true_num), round(false_num), round(num_false_flag), round(num_true_flag)

output = calc_nums(back, lin, true_split, false_split, number)
print(output)
print(f"fraction of false flag: {output[4]/(output[4]+output[5])}")
```
